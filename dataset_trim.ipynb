```python
# --- Configuration ---

# 1. Set your input and output folders
INPUT_FOLDER = "/content/drive/MyDrive/AI-Fiction/shobdotori"
OUTPUT_FOLDER = "/content/drive/MyDrive/AI-Fiction/trimed_shobdotori"

# 2. Set the silence threshold (in decibels)
SILENCE_THRESHOLD_DB = 50

# ... (rest of the script) ...
```

```python
import librosa
import soundfile as sf  # Used for saving the audio files
import os
import glob # Create the output directory if it doesn't exist
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# Find all .wav files in the input folder (including all subfolders)
search_path = os.path.join(INPUT_FOLDER, "**", "*.wav")

print(f"Starting batch trim from '{INPUT_FOLDER}'...")

for audio_file_path in glob.glob(search_path, recursive=True):
    # This check prevents re-processing files if you run it in the same folder
    if OUTPUT_FOLDER in audio_file_path:
        continue

    print(f"Processing: {audio_file_path}")

    try:
        # Load the audio file, preserving original sample rate
        y, sr = librosa.load(audio_file_path, sr=None)

        # 1. Find the start of the non-silent audio
        y_trimmed, index = librosa.effects.trim(y, top_db=SILENCE_THRESHOLD_DB)

        # 'index[0]' is the frame where the non-silent audio starts
        start_frame = index[0]

        # 2. Slice the *original* audio from that start frame to the end
        y_trimmed_start_only = y[start_frame:]

        # Note: If you want to trim silence from BOTH ends,
        # just use 'y_trimmed' instead of 'y_trimmed_start_only'

        # 3. Create the corresponding output path
        # This is the magic part that preserves your 'train/region/' structure
        relative_path = os.path.relpath(audio_file_path, INPUT_FOLDER)
        output_path = os.path.join(OUTPUT_FOLDER, relative_path)

        # Create the subfolder in the output directory if needed
        os.makedirs(os.path.dirname(output_path), exist_ok=True)

        # 4. Save the new audio file
        sf.write(output_path, y_trimmed_start_only, sr)

    except Exception as e:
        print(f"Error processing {audio_file_path}: {e}")

print("--- Batch trimming complete! ---")
print(f"Trimmed files are saved in: '{OUTPUT_FOLDER}'")
```
